## Setup Gemini API

To use the Gemini API, you'll need an API key. If you don't already have one, create a key in [Google AI Studio](https://makersuite.google.com/app/apikey).

In Colab, add the key to the secrets manager under the "🔑" icon in the left panel. Name it `GOOGLE_API_KEY`. Then, you can access it securely within your notebook.

```python
# Import the Python SDK
import google.generativeai as genai
# Used to securely store your API key
from google.colab import userdata

# Configure the Gemini API with your API key
GOOGLE_API_KEY=userdata.get('GOOGLE_API_KEY')
genai.configure(api_key=GOOGLE_API_KEY)
```

### Initialize the Generative Model

Now, let's initialize the Gemini Pro model, which is well-suited for chat-like interactions.

```python
# Initialize the Gemini Pro model
model = genai.GenerativeModel('gemini-pro')
```

## Create a Chat Interface

Now we can create a simple loop to take your input, send it to the Gemini model, and display the response. Type `exit` to end the chat.

```python
chat = model.start_chat(history=[])

print("Hello! I am Gemini. Type 'exit' to end our conversation.\n")

while True:
    user_input = input("You: ")
    if user_input.lower() == 'exit':
        print("Goodbye!")
        break

    try:
        response = chat.send_message(user_input)
        print(f"Gemini: {response.text}")
    except Exception as e:
        print(f"An error occurred: {e}")
```
